# Hissə 2: Vizuallaşdırma və Əlaqələrin Analizi
## - Datasetdə hansı aviaşirkətlər var və onların tezlikləri nə qədərdir?

```python
airline_frequencies = df['airline'].value_counts()
print("Datasetdəki aviaşirkətlər və uçuş tezlikləri:")
print(airline_frequencies)
```

## Gediş vaxtı və Gəliş vaxtını təmsil edən Bar Qrafikləri göstərin.

```python
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
df['departure_time'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Gediş Vaxtlarının Paylanması (Departure Time)')
plt.xlabel('Gediş Vaxtı')
plt.ylabel('Uçuş Sayı')
plt.xticks(rotation=45)
```

```python
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 2)
df['arrival_time'].value_counts().plot(kind='bar', color='salmon', edgecolor='black')
plt.title('Gəliş Vaxtlarının Paylanması (Arrival Time)')
plt.xlabel('Gəliş Vaxtı')
plt.ylabel('Uçuş Sayı')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

##  Mənbə Şəhəri və Təyinat Şəhəri təmsil edən Bar Qrafikləri göstərin.

```python
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
df['source_city'].value_counts().plot(kind='bar', color='teal', edgecolor='black')
plt.title('Mənbə Şəhərlərinin Paylanması (Source City)')
plt.xlabel('Mənbə Şəhər')
plt.ylabel('Uçuş Sayı')
plt.xticks(rotation=45)
```

```python
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 2)
df['destination_city'].value_counts().plot(kind='bar', color='orchid', edgecolor='black')
plt.title('Təyinat Şəhərlərinin Paylanması (Destination City)')
plt.xlabel('Təyinat Şəhər')
plt.ylabel('Uçuş Sayı')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## - Qiymət aviaşirkətlərə görə dəyişirmi?

```python
airline_prices = df.groupby('airline')['price'].mean().sort_values(ascending=False)
print("Aviaşirkətlərə görə orta bilet qiymətləri:")
print(airline_prices)
```

```python
plt.figure(figsize=(10, 5))
airline_prices.plot(kind='bar', color='darkorange', edgecolor='black')
plt.title('Aviaşirkətlərə görə Orta Bilet Qiymətləri')
plt.xlabel('Aviaşirkət')
plt.ylabel('Orta Qiymət')
plt.xticks(rotation=45)
plt.show()
```

## - Bilet qiyməti gediş vaxtı və gəliş vaxtına əsasən dəyişirmi?

```python
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
df.groupby('departure_time')['price'].mean().sort_values().plot(kind='bar', color='mediumseagreen', edgecolor='black')
plt.title('Gediş Vaxtına görə Orta Qiymət')
plt.xlabel('Gediş Vaxtı')
plt.ylabel('Orta Qiymət')
plt.xticks(rotation=45)
```

```python
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 2)
df.groupby('arrival_time')['price'].mean().sort_values().plot(kind='bar', color='indianred', edgecolor='black')
plt.title('Gəliş Vaxtına görə Orta Qiymət')
plt.xlabel('Gəliş Vaxtı')
plt.ylabel('Orta Qiymət')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## - Qiymət mənbə və təyinatın dəyişməsi ilə necə dəyişir?

```python
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
df.groupby('source_city')['price'].mean().sort_values().plot(kind='bar', color='royalblue', edgecolor='black')
plt.title('Mənbə Şəhərinə görə Orta Qiymət')
plt.xlabel('Mənbə Şəhər')
plt.ylabel('Orta Qiymət')
plt.xticks(rotation=45)
```

```python
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 2)
df.groupby('destination_city')['price'].mean().sort_values().plot(kind='bar', color='darkorchid', edgecolor='black')
plt.title('Təyinat Şəhərinə görə Orta Qiymət')
plt.xlabel('Təyinat Şəhər')
plt.ylabel('Orta Qiymət')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## - Biletlər uçuşdan cəmi 1 və ya 2 gün əvvəl alındıqda qiymət necə təsirlənir?

```python
df['buying_time'] = df['days_left'].apply(lambda x: '1-2 gün əvvəl' if x <= 2 else '3+ gün əvvəl')

early_vs_late_price = df.groupby('buying_time')['price'].mean()
print("Alınma vaxtına görə orta qiymətlər:")
print(early_vs_late_price)
```

```python
plt.figure(figsize=(8, 5))
early_vs_late_price.plot(kind='bar', color=['crimson', 'darkgray'], edgecolor='black')
plt.title('Son 1-2 Gündə Alınan Biletlərin Qiymət Müqayisəsi')
plt.xlabel('Biletin Alınma Vaxtı')
plt.ylabel('Orta Qiymət')
plt.xticks(rotation=0)
plt.show()
```

## - Bilet qiyməti İqtisadiyyat və Biznes sinfi arasında necə dəyişir?

```python
class_prices = df.groupby('class')['price'].mean()
print("Siniflərə görə (Economy/Business) orta qiymətlər:")
print(class_prices)
```

```python
plt.figure(figsize=(7, 5))
class_prices.plot(kind='bar', color=['navy', 'gold'], edgecolor='black')
plt.title('Uçuş Sinfinə görə Orta Qiymət fərqi')
plt.xlabel('Uçuş Sinfi')
plt.ylabel('Orta Qiymət')
plt.xticks(rotation=0)
plt.show()
```

## - Datasetdəki uçuşların orta müddəti (duration) nə qədərdir və bu müddət sinflərə (Economy/Business) görə necə fərqlənir?

```python
overall_mean_duration = df['duration'].mean()
print(f"Datasetdəki uçuşların ümumi orta müddəti: {overall_mean_duration:.2f} saat")
print("-" * 50)
```

```python
class_duration = df.groupby('class')['duration'].mean()
print("Siniflərə görə orta uçuş müddəti:")
print(class_duration)
```

```python
plt.figure(figsize=(7, 5))
class_duration.plot(kind='bar', color=['darkcyan', 'chocolate'], edgecolor='black')
plt.title('Siniflərə görə Orta Uçuş Müddəti (Duration)')
plt.xlabel('Uçuş Sinfi')
plt.ylabel('Orta Müddət (Saat)')
plt.xticks(rotation=0)
plt.show()
```

## - Durak sayı (stops) qiymətə necə təsir edir? (Məsələn, zero, one, two_or_more duraklar üçün orta qiymətləri müqayisə edin.)

```python
stops_prices = df.groupby('stops')['price'].mean().sort_values()
print("Durak sayına görə orta bilet qiymətləri:")
print(stops_prices)
```

```python
plt.figure(figsize=(8, 5))
stops_prices.plot(kind='bar', color='indigo', edgecolor='black')
plt.title('Durak Sayına (Stops) görə Orta Bilet Qiyməti')
plt.xlabel('Durak Sayı')
plt.ylabel('Orta Qiymət')
plt.xticks(rotation=0)
plt.show()
```

## - Günlər qalmış (days_left) dəyişkəninə görə qiymətin paylanması necədir? (Məsələn, 1-10 gün, 11-20 gün və s. qruplar üzrə qrafik və ya orta qiymətlər.)

```python
bins = [0, 10, 20, 30, 40, 50]
labels = ['1-10 gün', '11-20 gün', '21-30 gün', '31-40 gün', '41-50 gün']

df['days_left_group'] = pd.cut(df['days_left'], bins=bins, labels=labels)
```

```python
days_group_prices = df.groupby('days_left_group')['price'].mean()
print("Uçuşa qalan gün qruplarına görə orta qiymətlər:")
print(days_group_prices)
```

```python
plt.figure(figsize=(10, 5))
days_group_prices.plot(kind='bar', color='darkred', edgecolor='black')
plt.title('Uçuşa Qalan Gün Qruplarına görə Orta Bilet Qiyməti')
plt.xlabel('Qalan Gün Qrupları')
plt.ylabel('Orta Qiymət')
plt.xticks(rotation=0)
plt.show()
```

## - Hansı mənbə şəhərdən (source_city) ən çox uçuş var və onların orta qiyməti nə qədərdir?

```python
source_analysis = df.groupby('source_city').agg(
    flight_count=('price', 'count'),
    mean_price=('price', 'mean')
).sort_values(by='flight_count', ascending=False)

print("Mənbə şəhərlərinin analizi (Uçuş sayı və Orta qiymət):")
print(source_analysis)
```

```python
plt.figure(figsize=(9, 5))
source_analysis['flight_count'].plot(kind='bar', color='darkslateblue', edgecolor='black')
plt.title('Mənbə Şəhərlərinə görə Uçuş Sayları')
plt.xlabel('Mənbə Şəhər')
plt.ylabel('Uçuş Sayı')
plt.xticks(rotation=0)
plt.show()
```

## - Biznes sinfində uçuşların ümumi sayı və orta qiyməti İqtisadiyyat sinfi ilə müqayisədə necədir?

```python
class_comparison = df.groupby('class').agg(
    total_flights=('price', 'count'),
    average_price=('price', 'mean')
)
print("Siniflərin Müqayisə Cədvəli:")
print(class_comparison)
```
